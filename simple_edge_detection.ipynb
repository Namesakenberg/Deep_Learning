```python
import pandas as pd
import numpy as np
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
import cv2

```

```python
vertical = cv2.imread('/content/7b0001f5bb9ff63337746e7e9693d6f4.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
cv2_imshow(vertical)
```

```python
vertical.shape
```

```python
rows,cols = vertical.shape
```

```python
sobel_vertical = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])

print(sobel_vertical)
```

```python
kr, kc = sobel_vertical.shape
```

```python
output = np.zeros((rows-kr+1 , cols-kc+1))
```

```python
output.shape
```

```python
# iterate the kernal on the input
```

```python
for i in range(rows-kr+1):
  for j in range(cols-kc+1):
    curr_sum = 0
    for pi in range(kr):
      for pj in range(kc):
        curr_sum+=vertical[i+pi][j+pj] * sobel_vertical[pi][pj]
    output[i][j] = curr_sum


```

```python
cv2_imshow(output )
```

# using openCV

```python

output2 = np.zeros((rows-kr+1 , cols-kc+1))
```

```python
output2=cv2.filter2D(src = vertical , ddepth=-1 , kernel= sobel_vertical)
```

```python
cv2_imshow(output2)
```
